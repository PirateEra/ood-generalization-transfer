# Utils

```python
import matplotlib.pyplot as plt
import os
import json
import numpy as np

def set_plot_y_ticks(scores, num_ticks=15):
    max_score = max(scores)

    y_max = min(1, max_score + 0.01)

    ticks = np.linspace(0, y_max, num_ticks)
    ticks = np.round(ticks, 3)

    plt.yticks(ticks)


def to_sentence_case(s):
    return ' '.join(word.capitalize() for word in s.split('_')[1:])


def create_bar_plot(scores, metric, dataset, barplot=True):
    metric_name = to_sentence_case(metric)
    # Sort the dictionary for nicer plotting
    sorted_items = sorted(scores.items())
    models = [item[0] for item in sorted_items]
    values = [item[1] for item in sorted_items]

    # Plotting
    plt.figure(figsize=(12, 6))
    if barplot:
        plt.bar(models, values, color="skyblue", edgecolor="black")
    else:
        plt.plot(models, values, marker='o')
    set_plot_y_ticks(values, 15)
    plt.xticks(rotation=45, ha='right')
    plt.ylabel(f"{metric_name}")
    plt.title(f"Evaluation of {metric_name} on {dataset} Dataset")
    plt.tight_layout()
    plt.show()

def create_eval_dictionairy(path, metric):
    scores = {}
    for filename in os.listdir(path):
        if filename.endswith(".json"):
            file_path = os.path.join(path, filename)
            with open(file_path, "r") as f:
                data = json.load(f)
                # Get the scores
                value = data.get(metric)
                scores[filename.split("_")[0]] = value
    return scores
```

# Evaluation

```python
dataset = "CancerEmo"
folder_path = f"eval_results/{dataset}"
metric = "eval_f1_macro"
scores = create_eval_dictionairy(folder_path, metric)
create_bar_plot(scores, metric, dataset)
```
